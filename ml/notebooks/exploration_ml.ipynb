```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve().parents[1] if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
import pandas as pd
from ml.data_generation import generate_dataset
from ml.config import DEFAULT_DATASET_PATH, NUMERIC_COLUMNS
from ml.preprocessing import make_feature_frame
if not DEFAULT_DATASET_PATH.exists():
    generate_dataset(DEFAULT_DATASET_PATH)
data = pd.read_csv(DEFAULT_DATASET_PATH)
data.head()
```

```python
display(data['risk_level'].value_counts().rename_axis('risk_level').to_frame('count'))
data[NUMERIC_COLUMNS].describe().T
```
